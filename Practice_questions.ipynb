```python
#11
def primes_upto(n):
    primes = []
    for num in range(2, n+1):
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                break
        else:
            primes.append(num)
    return primes

print(primes_upto(50))
```

```python
#12
def fibonacci(n):
    a, b = 0, 1
    seq = []
    for _ in range(n):
        seq.append(a)
        a, b = b, a+b
    return seq

print(fibonacci(10))
```

```python
#17
def remove_duplicates(lst):
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            result.append(item)
            seen.add(item)
    return result

print(remove_duplicates([1,2,2,3,4,4,5]))
```

```python
#14
def sort_list(lst):
    for i in range(len(lst)):
        for j in range(i+1, len(lst)):
            if lst[i] > lst[j]:
                lst[i], lst[j] = lst[j], lst[i]
    return lst

print(sort_list([5,2,9,1,5,6]))
```

```python
#5
def tower_of_hanoi(n, source, target, auxiliary):
    if n == 1:
        print(f"Move disk 1 from {source} to {target}")
        return
    tower_of_hanoi(n-1, source, auxiliary, target)
    print(f"Move disk {n} from {source} to {target}")
    tower_of_hanoi(n-1, auxiliary, target, source)
tower_of_hanoi(3, "A", "C", "B")
```

```python
#8
def read_numbers():
    numbers = []
    while True:
        user_input = input("Enter a number (or 'stop' to finish): ")
        if user_input.lower() == "stop":
            break
        try:
            num = float(user_input)  
            numbers.append(num)
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    return numbers

def calculate_stats(numbers):
    if not numbers:
        print("No numbers entered.")
        return

    minimum = min(numbers)
    maximum = max(numbers)
    average = sum(numbers) / len(numbers)

    sorted_nums = sorted(numbers)
    n = len(sorted_nums)
    if n % 2 == 1:
        median = sorted_nums[n // 2]
    else:
        median = (sorted_nums[n // 2 - 1] + sorted_nums[n // 2]) / 2

    print(f"Numbers: {numbers}")
    print(f"Minimum: {minimum}")
    print(f"Maximum: {maximum}")
    print(f"Average: {average:.2f}")
    print(f"Median: {median}")
numbers = read_numbers()
calculate_stats(numbers)
```

```python
#9
def flatten_list(nested_list):
    if not nested_list:
        return []
    if isinstance(nested_list[0], list):
        return flatten_list(nested_list[0]) + flatten_list(nested_list[1:])
    else:
        return [nested_list[0]] + flatten_list(nested_list[1:])
nested = [1, [2, [3, 4], 5], [6, 7], 8]
print(flatten_list(nested))  
```

```python
#10
import csv
students = []
def add_student(student_id, name, marks):
    try:
        marks = float(marks)
        students.append({"ID": student_id, "Name": name, "Marks": marks})
        print("Student added successfully.")
    except ValueError:
        print("Marks must be a number.")
def search_student(student_id):
    for s in students:
        if s["ID"] == student_id:
            print("Found:", s)
            return s
    print("Student not found.")
    return None
def update_marks(student_id, new_marks):
    for s in students:
        if s["ID"] == student_id:
            try:
                s["Marks"] = float(new_marks)
                print("Marks updated.")
                return
            except ValueError:
                print("Invalid marks.")
                return
    print("Student not found.")
def delete_student(student_id):
    global students
    students = [s for s in students if s["ID"] != student_id]
    print("Deleted if existed.")
def show_stats():
    if not students:
        print("No records.")
        return
    marks = [s["Marks"] for s in students]
    print("Min:", min(marks))
    print("Max:", max(marks))
    print("Average:", sum(marks)/len(marks))
def save_data(filename="students.csv"):
    if students:
        with open(filename, "w", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=["ID","Name","Marks"])
            writer.writeheader()
            writer.writerows(students)
        print("Data saved.")
    else:
        print("No data to save.")
def load_data(filename="students.csv"):
    global students
    try:
        with open(filename, "r") as f:
            reader = csv.DictReader(f)
            students = [row for row in reader]
            for s in students:
                s["Marks"] = float(s["Marks"])
        print("Data loaded.")
    except FileNotFoundError:
        print("File not found, starting fresh.")
```

```python
add_student("101", "Priya", 95)
add_student("102", "Amit", 88)

search_student("101")
update_marks("102", 90)

show_stats()
delete_student("101")

save_data()
load_data()
```
